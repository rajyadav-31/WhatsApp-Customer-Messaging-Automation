```python
# This is for multiple Peoples and fully Automatic
import pyautogui
import time

pyautogui.click(424, 1049)
pyautogui.write("whatsapp", 0.2)
pyautogui.press("enter")

time.sleep(3)

for name in l:
    pyautogui.click(380, 252)  # Search box
    pyautogui.write(name, 0.2)
    time.sleep(2)

    pyautogui.click(400, 485)  # Select chat
    time.sleep(2)

    pyautogui.write("Heyy!!")
    pyautogui.sleep(3)
    pyautogui.press("enter")

    time.sleep(2)

# khatam..!!!!
```
